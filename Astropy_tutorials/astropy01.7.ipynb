There are a few choices you need to make about how you will calibrate your data. Sometimes the decision will be made for you by the data you have.

# Subtract bias and dark as separate steps or in one step?

Every dark image contains bias when it comes off of the camera, so in principle you can take care of both bias and dark by constructing a master dark image that leaves the bias already present in the dark images in place.

This only works if for every image from which you need to remove dark currrent you have a master dark of exactly the same exposure length.

If not, you need to produce master dark images with the bias removed so that they can be scaled by exposure time.

*Se todos os darks têm o mesmo tempo de exposência das imagens de ciência → fazer junto. Caso contrário → separar bias e dark.*

# Subtract overscan or not?

This is only applicable if your images have an <u>overscan region.</u> Subtracting overscan can be useful for removing small (typically a few counts) variations from image-to-image over the course of a night.

*Overscan é útil se a região for confiável e precisa corrigir variações pequenas de bias frame a frame. Caso a camêra <u>não</u> tenha overscan ou for um overscan ruim (como o exemplo 2 no notebook anterior), nem tenta !!!*

<h1 style="text-align: left; font-family: 'menlo'; color: #191970; font-size:40px;">
  <span style="background-color: #ADD8E6; padding: 5px 10px; border-radius: 5px; display: inline-block;">
     Notes
  </span>
</h1>

- Fazer um master dark com o EXATO mesmo tempo de exposição da sua imagem de science → combina os darks (sem remover o bias) → subtrai esse master dark da imagem de science → removidos dark + bias de uma vez.

- Se não tem um dark com o mesmo tempo de exposição da science, precisa escalar o dark pelo tempo de exposição. Pra isso, o dark precisa estar limpo de bias:

    - `dark limpo = dark original − bias`
    - então: `dark escalado = dark limpo × (tempo_science / tempo_dark)`
    - por fim, subtratir: `imagem calibrada = science − bias − dark_escalado`
