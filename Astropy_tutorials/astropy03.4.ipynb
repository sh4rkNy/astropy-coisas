Dark images, like any other images, need to be calibrated. Depending on the data you have and the choices you have made in reducing your data, the steps to reducing your images may include:

- <u>Subtracting overscan</u> (only if you decide to subtract overscan from all images).

- <u>Trim the image</u> (if it has overscan, whether you are using the overscan or not).

- <u>Subtract bias</u> (if you need to scale the calibrated dark frames to a different exposure time).

```python
from pathlib import Path

from astropy.nddata import CCDData
from ccdproc import ImageFileCollection
import ccdproc as ccdp
```

# Example 1: Overscan subtracted, bias not removed

## Take a look at what images you have

First we gather up some information about the raw images and the reduced images up to this point. These examples have darks stored in a subdirectory of the folder with the rest of the images, so we create an <span style="color: #d62495">ImageFileCollection</span> for each.

```python
ex1_path_raw = Path('example-cryo-LFC')

ex1_images_raw = ImageFileCollection(ex1_path_raw)
ex1_darks_raw = ImageFileCollection(ex1_path_raw / 'darks')

ex1_path_reduced = Path('example1-reduced')
ex1_images_reduced = ImageFileCollection(ex1_path_reduced)
```

## Raw images, everything except the darks

```python
ex1_images_raw.summary['file', 'imagetyp', 'exptime', 'filter']
```

## Raw dark frames

```python
ex1_darks_raw.summary['file', 'imagetyp', 'exptime', 'filter']
```

## Decide which calibration steps to take

This example is, again, one of the chips of the LFC camera at Palomar. In earlier notebooks we have seen that the chip has a [useful overscan region](https://www.astropy.org/ccd-reduction-and-photometry-guide/v/dev/notebooks/01-08-Overscan.ipynb#case-1-cryogenically-cooled-large-format-camera-lfc-at-palomar), <u>has little dark current except for some hot pixels, and sensor glow in one corner of the chip.</u>

Looking at the list of non-dark images (i.e., the flat and light images) shows that for each exposure time in the non-dark images there is a set of dark exposures that has a matching, or very close to matching, exposure time.

To be more explicit, there are flats with exposure times of 7.0 sec and 70.011 sec and darks with exposure time of 7.0 and 70.0 sec. <u>The dark and flat exposure times are close enough that there is no need to scale them.</u> The two images of an object are each roughly 300 sec, matching the darks with exposure time 300 sec. The very small difference in exposure time, under 0.1 sec, does not need to be compensated for.

Given this, we will:

1. <u>Subtract overscan</u> from each of the darks. The useful overscan region is XXX (see LINK).

2. <u>Trim the overscan</u> out of the dark images.

We will **not** subtract bias from these images because we will not need to rescale them to a different exposure time.

```python
for ccd, file_name in ex1_darks_raw.ccds(imagetyp='DARK',
                                        ccd_kwargs={'unit': 'adu'},
                                        return_fname=True):
    ccd = ccdp.subtract_overscan(ccd, overscan=ccd[:, 2055:], median=True)
    ccd = ccdp.trim_image(ccd[:, :2048])
    ccd.write(ex1_path_reduced / file_name) #save
```

### Reduced images (so far)

```python
ex1_images_reduced.refresh()
ex1_images_reduced.summary['file', 'imagetyp', 'exptime', 'filter', 'combined']
```

# Example 2: Overscan not subtracted, bias is removed

```python
ex2_path_raw = Path('example-thermo-electric')
ex2_images_raw = ImageFileCollection(ex2_path_raw)

ex2_path_reduced = Path('example2-reduced')
ex2_images_reduced = ImageFileCollection(ex2_path_reduced)
```

We begin by looking at what exposure times we have in this data.

```python
ex2_images_raw.summary['file', 'imagetyp', 'exposure'].show_in_notebook()
```

##  Decide what steps to take next

In this case the only dark frames have exposure time 90 sec. Though that matches the exposure time of the science images, the <u>flat field images</u> are much shorter exposure time, ranging from 1 sec to 1.21 sec. This type of range of exposure is typical when twilight flats are taken. Since these are a much different exposure time than the darks, the dark frames will need to be scaled.

Recall that for this camera the overscan is not useful and should be trimmed off.

Given this, we will:

1. Trim the overscan from each of the dark frames.

2. Subtract calibration bias from the dark frames so that we can scale the darks to a different exposure time.

## Calibration the individual dark frames

<u>First, we read the combined bias image created in the previous notebook.</u> Though we could do this based on the file name, using a systematic set of header keywords to keep track of which images have been combined is less likely to lead to errors.

```python
combined_bias = CCDData.read(ex2_images_reduced.files_filtered(imagetyp='bias',
                                                              combined=True,
                                                              include_path=True)[0])
```

```python
for ccd, file_name in ex2_images_raw.ccds(imagetyp='DARK',
                                         return_fname=True):
    ccd = ccdp.trim_image(ccd[:, :4096])
    ccd = ccdp.subtract_bias(ccd, combined_bias)
    ccd.write(ex2_path_reduced / file_name)
```
